**Objectif :**

Utiliser l'algorithme **YOLO (You Only Look Once)** pour détecter les objets dans une image.

La détection d'objets fait deux choses :
1. **Localiser** les objets (dessiner une boîte autour)
2. **Classifier** (dire ce que c'est : chat, chien, personne...)

## Étape 1 — Installation

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import urllib.request

print("Prêt !")
```

---
## Étape 2 — Charger le modèle YOLOv8

```python
model = YOLO('yolov8n.pt')
print(f"Modèle chargé. Il détecte {len(model.names)} classes.")
print(f"Exemples : {list(model.names.values())[:10]}")
```

---
## Étape 3 — Choisir une image

**Option 1** : Télécharger une image depuis une URL

**Option 2** : Téléverser votre propre image

### Option 1 — Télécharger depuis une URL

```python
# Modifier l'URL pour choisir une autre image
url = "https://ultralytics.com/images/bus.jpg"

urllib.request.urlretrieve(url, "image.jpg")
print("Image téléchargée !")
```

### Option 2 — Téléverser votre propre image

Exécutez la cellule ci-dessous, puis glissez-déposez votre image.

```python
# Téléverser une image depuis votre ordinateur
from google.colab import files

print("Choisissez une image :")
uploaded = files.upload()

# Récupérer le nom du fichier
filename = list(uploaded.keys())[0]

# Renommer pour simplifier
import os
os.rename(filename, "image.jpg")

print(f"\nImage '{filename}' chargée !")
```

---
## Étape 4 — Afficher l'image

```python
img = cv2.imread("image.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 7))
plt.imshow(img_rgb)
plt.axis('off')
plt.title('Image choisie')
plt.show()

print(f"Taille : {img.shape[1]}x{img.shape[0]} pixels")
```

---
## Étape 5 — Lancer la détection

```python
# Détecter les objets
results = model("image.jpg")

# Afficher avec les boîtes
annotated = results[0].plot()

plt.figure(figsize=(12, 8))
plt.imshow(annotated[:, :, ::-1])
plt.axis('off')
plt.title('Objets détectés')
plt.show()
```

---
## Étape 6 — Voir les résultats en détail

```python
boxes = results[0].boxes

print(f"Nombre d'objets détectés : {len(boxes)}")
print()

for i, box in enumerate(boxes, 1):
    classe = model.names[int(box.cls[0])]
    confiance = float(box.conf[0])
    xyxy = box.xyxy[0].tolist()
    
    print(f"{i}. {classe} — confiance : {confiance:.0%}")
    print(f"   Position : ({xyxy[0]:.0f}, {xyxy[1]:.0f}) → ({xyxy[2]:.0f}, {xyxy[3]:.0f})")
    print()
```

---
## Étape 7 — Sauvegarder

```python
cv2.imwrite("resultat.jpg", annotated)
print("Image sauvegardée : resultat.jpg")
```

### Comment ça marche ?
1. L'image est découpée en grille
2. Chaque cellule prédit des boîtes + classes
3. Le post-traitement (NMS) élimine les doublons

Le modèle COCO détecte 80 classes : personnes, voitures, animaux, objets du quotidien...
