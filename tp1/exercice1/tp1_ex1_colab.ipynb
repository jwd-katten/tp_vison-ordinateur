## Étape 3 — Afficher l'image couleur

Une image couleur a 3 canaux : **B** (Bleu), **G** (Vert), **R** (Rouge).

Chaque pixel = un vecteur [B, G, R] avec des valeurs de 0 à 255.

```python
# Lire l'image avec OpenCV (charge en BGR par défaut)
img = cv2.imread("image.png")

# Convertir BGR → RGB pour l'affichage avec matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.axis('off')
plt.title('Image couleur (RGB)')
plt.show()

print(f"Taille : {img.shape}")
print(f"Nombre de canaux : {img.shape[2]}")
```

## Étape 4 — Convertir en niveaux de gris

La conversion utilise la formule :

$$\text{Gris} = 0.299 \times R + 0.587 \times G + 0.114 \times B$$

Cette formule pondère les canaux selon la sensibilité de l'œil humain.

```python
# Convertir en niveaux de gris
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print(f"Taille en gris : {gray.shape}")
print(f"Valeurs : min={gray.min()}, max={gray.max()}")
```

## Étape 5 — Comparer les deux images

```python
# Afficher côte à côte
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].imshow(img_rgb)
axes[0].set_title('Couleur (RGB)', fontsize=14)
axes[0].axis('off')

axes[1].imshow(gray, cmap='gray')
axes[1].set_title('Niveaux de gris', fontsize=14)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Étape 6 — Sauvegarder le résultat

```python
# Sauvegarder l'image en gris
cv2.imwrite("image_gris.jpg", gray)
print("Image sauvegardée : image_gris.jpg")
```

## Résumé

| Étape | Code |
|-------|------|
| Charger | `cv2.imread('image.jpg')` |
| Convertir | `cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)` |
| Sauvegarder | `cv2.imwrite('output.jpg', gray)` |

**À retenir :**
- OpenCV charge en BGR, pas RGB
- Le gris a 1 seul canal (vs 3 pour la couleur)
- La conversion réduit la complexité tout en gardant l'information
