```python
import cv2
import matplotlib.pyplot as plt

print("Prêt !")
```

## Étape 2 — Charger une image

```python
import urllib.request

url = "https://ultralytics.com/images/bus.jpg"
urllib.request.urlretrieve(url, "image.jpg")

img = cv2.imread("image.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 6))
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.title(f'Image originale ({gray.shape[1]}x{gray.shape[0]})')
plt.show()
```

---
## Étape 3 — Filtre médian

Le filtre médian remplace chaque pixel par la **médiane** de son voisinage.

**Différence avec le gaussien :**
- Gaussien : moyenne pondérée → flou
- Médian : médiane → préserve mieux les bords

**Idéal pour :** le bruit "poivre et sel" (pixels blancs/noirs aléatoires)

```python
# Filtre médian (taille 5)
median = cv2.medianBlur(gray, 5)

# Filtre gaussien pour comparaison
gauss = cv2.GaussianBlur(gray, (5, 5), 0)

# Comparer
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(gray, cmap='gray')
axes[0].set_title('Original', fontsize=14)
axes[0].axis('off')

axes[1].imshow(gauss, cmap='gray')
axes[1].set_title('Gaussien (5x5)', fontsize=14)
axes[1].axis('off')

axes[2].imshow(median, cmap='gray')
axes[2].set_title('Médian (5)', fontsize=14)
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

---
## Étape 4 — Redimensionnement

Redimensionner change la taille d'une image.

**Interpolation :**
- `cv2.INTER_LINEAR` : rapide, bon par défaut
- `cv2.INTER_CUBIC` : plus lent, meilleure qualité
- `cv2.INTER_LANCZOS4` : meilleure qualité

```python
# Redimensionner à 320x240
resized = cv2.resize(median, (320, 240))

print(f"Taille originale : {median.shape[1]}x{median.shape[0]}")
print(f"Taille après redimensionnement : {resized.shape[1]}x{resized.shape[0]}")

# Afficher
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].imshow(median, cmap='gray')
axes[0].set_title(f'Original ({median.shape[1]}x{median.shape[0]})', fontsize=14)
axes[0].axis('off')

axes[1].imshow(resized, cmap='gray')
axes[1].set_title(f'Redimensionné (320x240)', fontsize=14)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Étape 5 — Sauvegarder

```python
cv2.imwrite("image_median.jpg", median)
cv2.imwrite("image_resized.jpg", resized)

print("Images sauvegardées !")
```

## Résumé

| Filtre | Avantage | Inconvénient |
|--------|----------|---------------|
| **Gaussien** | Rapide, lisse bien | Flou les contours |
| **Médian** | Préserve les bords | Plus lent |

| Opération | Code |
|-----------|------|
| Médian | `cv2.medianBlur(img, 5)` |
| Redimensionner | `cv2.resize(img, (largeur, hauteur))` |
