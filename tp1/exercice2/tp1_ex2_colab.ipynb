```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

print("Prêt !")
```

## Étape 2 — Charger une image

```python
# Télécharger une image
import urllib.request

url = "https://ultralytics.com/images/bus.jpg"
urllib.request.urlretrieve(url, "image.jpg")

# Lire l'image avec OpenCV
img = cv2.imread("image.jpg")

# Convertir en niveaux de gris
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Afficher l'image en gris
plt.figure(figsize=(8, 6))
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.title('Image en niveaux de gris')
plt.show()
```

---
## Étape 3 — Filtrage gaussien

Le filtre gaussien lisse l'image en réduisant le bruit.

**Comment ?** Chaque pixel devient la moyenne pondérée de ses voisins (noyau gaussien).

**Paramètre :** `(5, 5)` = taille du noyau. Plus c'est grand, plus c'est flou.

```python
# Appliquer le filtre gaussien
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Comparer l'original et le flou
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].imshow(gray, cmap='gray')
axes[0].set_title('Original', fontsize=14)
axes[0].axis('off')

axes[1].imshow(blur, cmap='gray')
axes[1].set_title('Filtre gaussien (5x5)', fontsize=14)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Étape 4 — Histogramme

L'histogramme montre la distribution des valeurs de pixels :
- **Gauche (0)** = pixels sombres
- **Droite (255)** = pixels clairs

C'est utile pour analyser le contraste et la luminosité.

```python
# Calculer et afficher l'histogramme
plt.figure(figsize=(10, 5))
plt.hist(gray.ravel(), 256, [0, 256])
plt.title('Histogramme des niveaux de gris', fontsize=14)
plt.xlabel('Valeur de pixel')
plt.ylabel('Nombre de pixels')
plt.grid(True, alpha=0.3)
plt.show()
```

---
## Étape 5 — Détection de contours (Canny)

L'algorithme de Canny détecte les contours dans une image.

**Comment ça marche ?**
1. L'image est d'abord lissée avec un filtre gaussien
2. On calcule les changements d'intensité (les gradients)
3. On garde uniquement les points où le changement est fort

**Paramètres :** `seuil_bas` et `seuil_haut`
- Seuil bas : les contours faibles sont ignorés
- Seuil haut : les contours forts sont gardés

```python
# Appliquer Canny sur l'image en gris
edges = cv2.Canny(gray, 100, 200)

# Comparer les 3 images
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].imshow(gray, cmap='gray')
axes[0].set_title('Original en gris', fontsize=14)
axes[0].axis('off')

axes[1].imshow(blur, cmap='gray')
axes[1].set_title('Après filtre gaussien', fontsize=14)
axes[1].axis('off')

axes[2].imshow(edges, cmap='gray')
axes[2].set_title('Contours Canny (100, 200)', fontsize=14)
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

## Étape 6 — Comparaison complète

Affichons toutes les étapes côte à côte : couleur, gris, flou, contours, histogramme.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Image couleur
axes[0, 0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0, 0].set_title('Couleur', fontsize=14)
axes[0, 0].axis('off')

# Niveaux de gris
axes[0, 1].imshow(gray, cmap='gray')
axes[0, 1].set_title('Gris', fontsize=14)
axes[0, 1].axis('off')

# Filtre gaussien
axes[1, 0].imshow(blur, cmap='gray')
axes[1, 0].set_title('Gaussien (5x5)', fontsize=14)
axes[1, 0].axis('off')

# Contours Canny
axes[1, 1].imshow(edges, cmap='gray')
axes[1, 1].set_title('Contours Canny', fontsize=14)
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```

## Résumé

| Opération | Code |

| Histogramme | `plt.hist(img.ravel(), 256, [0,256])` |
| Contours Canny | `cv2.Canny(img, 100, 200)` |

**À retenir :**
